## Bevezetés

A mesterséges intelligencia, a mélytanulás (*deep learning*) és a neurális hálózatok a gépi tanulás fontos, számos gyakorlati feladatban alkalmazott módszerei. Az idegrendszer felépítése által ihletett neurális hálózatokat széles körben használják képfeldolgozásban és más mintafelismerési feladatokban.

Egy neurális hálózat képes nemlineáris kapcsolatokat közelíteni a bemenetek és a kimenetek között. A feldolgozás összekapcsolt neuronokon keresztül történik, a kapcsolatokhoz pedig módosítható súlyok tartoznak. A súlyokat a **tanítás** (*training*) során optimalizáljuk; ezt követően a megtanult paraméterekkel új adatokon **következtetést** (*inference*) végezhetünk.

## A neurális hálózatok kvantálása

A hálózatok súlyait hagyományosan 32 bites lebegőpontos számokkal ábrázolják. A kvantálással foglalkozó kutatások szerint bizonyos architektúrákban és feladatokban 8, 4, 2, sőt akár 1 bites súlyábrázolás is alkalmazható elfogadható pontosság mellett, különösen akkor, ha a kvantálást már a tanítás során figyelembe vesszük.

A 32 bites ábrázoláshoz képest a kisebb bitszélesség jelentősen csökkenti az ábrázolható értéktartományt. Ezt a kompromisszumot a predikciós pontosság, a memóriaigény és a hardvermegvalósítás költsége szempontjából kell értékelni.

A legfontosabb előnyök:

- **Kisebb memóriaigény:** a csökkentett pontosságú paraméterek tárolása kevesebb memóriát igényel.
- **Hatékonyabb aritmetika:** a kis bitszélességű egész- és fixpontos műveletek kevesebb hardvererőforrást és energiát igényelhetnek, mint a lebegőpontos számítások.
- **Az FPGA rugalmassága:** a programozható logika lehetővé teszi a kvantált adatszélességhez illesztett, célzott számítóegységek megvalósítását.

## Hálózattípusok és overlay-k

Az eredeti PYNQ-példák két fő csoportba sorolhatók:

- **BNN (Binarized Neural Network):** olyan hálózat, amelyben egyes súlyok és/vagy aktivációk bináris, egybites reprezentációt használnak. Megfelelő bináris operandusok esetén a szorzás–akkumulációs műveletek XNOR és *population count* (popcount) műveletekkel helyettesíthetők.
- **QNN (Quantized Neural Network):** kvantált súlyokkal és aktivációkkal működő hálózat, konfigurálható bitszélességekkel. A QNN-MO-PYNQ projekt többek között W1A2 és W1A3 konfigurációkat használ.

A másik lényeges különbség a **gyorsítóarchitektúra**:

### Feed-forward Dataflow

A hálózat minden rétege hardverben valósul meg. Egy réteg kimenete közvetlenül a következő réteg bemenete, amely az adatok megérkezésekor megkezdheti a feldolgozást. A paramétereket a chipen lévő memória tárolja. A megvalósítás egy adott hálózati topológiára specializált, így kis késleltetés és nagy áteresztőképesség érhető el, a rugalmasság rovására.

### Multi-layer Offload (MO)

Egy újrafelhasználható hardvergyorsító több hálózati réteget egymást követő hívások során hajt végre. A szoftver ütemezi a rétegek végrehajtását és kezeli az adatokat. A hálózati topológia változása a futtatás ütemezését módosíthatja anélkül, hogy a teljes hardverarchitektúrát újra kellene tervezni. Nagyobb rugalmasságot kapunk, de a vezérlés és adatmozgatás többletköltségével.

Az eredeti megoldásokban a **BNN** példák jellemzően dataflow-architektúrát alkalmaznak, míg a **QNN-MO** példák a gyorsító újrafelhasználására épülnek, 2 vagy 3 bites aktivációkkal.
