# 01. Introducere în învățarea automată pe PYNQ

*Material didactic: Prof. dr. ing. habil. Stefan Oniga. Bazat pe documentația și framework-ul original [Xilinx/QNN-MO-PYNQ](https://github.com/Xilinx/QNN-MO-PYNQ). Componentele originale își păstrează creditele și licențele.*

## Introducere

Inteligența artificială, învățarea profundă (*deep learning*) și rețelele neuronale sunt metode importante de învățare automată, utilizate pentru rezolvarea unor probleme reale. Inspirate de organizarea sistemului nervos, rețelele neuronale sunt folosite pe scară largă în prelucrarea imaginilor și în alte aplicații de recunoaștere a tiparelor.

O rețea neuronală poate aproxima relații neliniare între intrări și ieșiri. Informația este prelucrată prin neuroni interconectați, iar conexiunile conțin ponderi ajustabile. Ponderile sunt optimizate în etapa de **antrenare**; ulterior, cu parametrii învățați, rețeaua efectuează **inferență** asupra unor date noi.

## Cuantizarea rețelelor neuronale

În mod tradițional, ponderile unei rețele sunt reprezentate folosind numere în virgulă mobilă pe 32 de biți. Cercetările asupra cuantizării au arătat că, pentru anumite arhitecturi și sarcini, ponderile pot fi reprezentate cu 8, 4, 2 sau chiar 1 bit, fără o degradare inacceptabilă a acurateței, mai ales când cuantizarea este luată în considerare la antrenare.

Comparativ cu reprezentarea pe 32 de biți, reducerea preciziei micșorează semnificativ domeniul valorilor reprezentabile. Acest compromis trebuie evaluat în raport cu precizia predicțiilor, consumul de memorie și costul implementării hardware.

Avantajele principale sunt:

- **Memorie redusă:** reprezentările cu precizie mică ocupă mai puțin spațiu pentru stocarea milioanelor de parametri.
- **Aritmetică eficientă:** operațiile cu numere întregi sau în virgulă fixă de precizie redusă pot necesita mai puține resurse hardware și energie decât cele în virgulă mobilă.
- **Flexibilitatea FPGA:** logica programabilă permite realizarea unor unități de calcul personalizate pentru lățimile de date ale rețelei cuantizate.

## Tipuri de rețele și overlay-uri

În exemplele originale PYNQ întâlnim două categorii:

- **BNN (Binarized Neural Network):** rețele în care reprezentarea anumitor ponderi și/sau activări este binară, pe un singur bit. Pentru operanzi binari corespunzători, operațiile de înmulțire–acumulare pot fi implementate cu XNOR și *population count* (popcount).
- **QNN (Quantized Neural Network):** rețele cu ponderi și activări cuantizate, cu lățimi de date configurabile. În proiectul QNN-MO-PYNQ sunt utilizate, între altele, configurațiile W1A2 și W1A3.

O altă deosebire esențială este **arhitectura de accelerare**:

### Feed-forward Dataflow

Fiecare strat al rețelei este implementat în hardware; ieșirea unui strat alimentează direct stratul următor, care poate începe prelucrarea imediat ce datele devin disponibile. Parametrii sunt păstrați în memoria internă a circuitului, iar arhitectura este specializată pentru topologia rețelei. Această abordare poate oferi latență mică și debit ridicat, cu un cost de flexibilitate.

### Multi-layer Offload (MO)

Un accelerator hardware reutilizabil execută mai multe straturi ale rețelei în apeluri succesive. Între apeluri, software-ul stabilește ordinea operațiilor și gestionează datele. Schimbarea topologiei poate necesita modificarea programării execuției, fără reproiectarea întregii arhitecturi hardware. Rezultă o flexibilitate mai mare, dar și costuri suplimentare de control și transfer.

În implementările originale, exemplele **BNN** folosesc predominant arhitecturi de tip dataflow, în timp ce exemplele **QNN-MO** urmăresc reutilizarea acceleratorului pentru activări cuantizate pe 2 sau 3 biți.
